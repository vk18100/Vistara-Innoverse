```python
import pandas as pd
```

```python
df = pd.read_csv('C:\\Users\\INFINIX\\Vistara-Innoverse\\ml\\recommendation\\properties.csv')
df.head()
```

```python
verification_columns = ['property_id','property_name','verified','smoke_alarm','carbon_monoxide_alarm']
df[verification_columns].head(10)
```

```python
def verification_score(property):
    score = 0

    #For Smoke Alarm Score
    if property['smoke_alarm'] == 1:
        score += 0.30

    #For Carbon Monoxide Alarm
    if property['carbon_monoxide_alarm'] == 1:
        score += 0.30

    #For Verified Score
    if property['verified'] ==1:
        score += 0.40

    return score
```

```python
df['verification_score'] = df.apply(verification_score, axis = 1)

df[['property_id', 'property_name', 'verified', 'verification_score']]
```

```python
def verification_status(score):
    if score >= 0.8:
        return 'Verified'
    elif score >= 0.5:
        return 'Needs  Review'
    else:
        return 'Rejected'
```

```python
df["verification_status"] = df["verification_score"].apply(verification_status)

df[["property_id", "property_name","verification_score", "verification_status"]].head(10)
```

```python
df["verification_status"].value_counts()
```

```python
df[["property_id","property_name","verification_score","verification_status"]].head(10)
```

```python
def verification_reasons(property):
    reasons = []

    if property["verified"] == 1:
        reasons.append("Property information verified")

    if property["smoke_alarm"] == 1:
        reasons.append("Smoke alarm available")

    if property["carbon_monoxide_alarm"] == 1:
        reasons.append("Carbon monoxide alarm available")

    if property["verified"] == 0:
        reasons.append("Property verification not completed")

    if property["smoke_alarm"] == 0:
        reasons.append("Smoke alarm not confirmed")

    if property["carbon_monoxide_alarm"] == 0:
        reasons.append("Carbon monoxide alarm not confirmed")

    return reasons
```

```python
top_property = df.iloc[0]

verification_reasons(top_property)
```

```python
df["verification_status"] = df["verification_score"].apply(verification_status)
```

```python
df[["property_id","property_name","verification_score","verification_status"]].head(10)
```

```python
property = df.iloc[0]

print("Property:", property["property_name"])
print("Score:", property["verification_score"])
print("Status:", property["verification_status"])
```

```python
review_property = df[df["verification_status"] == "Needs  Review"].iloc[0]

print(review_property["property_name"])
print(review_property["verification_score"])
print(review_property["verification_status"])
```

```python
review_reasons = verification_reasons(review_property)

print("Property:", review_property["property_name"])
print("Reasons:")
for reason in review_reasons:
    print("✓", reason)
```

```python
print("Verification Score:", round(review_property["verification_score"] * 100, 2), "/ 100")
print("Status:", review_property["verification_status"])
```
