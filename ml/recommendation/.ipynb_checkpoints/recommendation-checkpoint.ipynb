```python
# Recommendatiton
user = {
    "city": "Delhi",
    "budget": 5000,
    "guests": 2,
    "bedroom": 4,
    "bathrooms": 1,

    "property_type": "Entire Home",

    "kitchen": 1,
    "wifi": 1,
    "parking": 1,
    "ac": 1,
    "balcony": 1,
    "terrace": 0,
    "workspace": 1,
    "pool": 0,
    "washing_machine": 1,

    "self_check_in": 1,
    "free_cancellation": 1,
    "allows_pets": 0,
    "check_in": "2026-11-01",
    "check_out": "2026-11-03"
}
```

```python
filters = {
    "property_type": user["property_type"],
    "bedrooms": user["bedroom"],
    "bathrooms": user["bathrooms"],
    "kitchen": user["kitchen"],
    "wifi": user["wifi"],
    "parking": user["parking"],
    "ac": user["ac"],
    "balcony": user["balcony"],
    "self_check_in": user["self_check_in"]
}
```

```python
filtered_df = df[
    (df["city"] == user["city"]) &
    (df["max_guests"] >= user["guests"]) &
    (df["available"] == 1) &
    (pd.to_datetime(df["available_from"]) <= user["check_in"]) &
    (pd.to_datetime(df["available_to"]) >= user["check_out"])
].copy()
```

```python
filtered_df = filtered_df[
    (filtered_df["property_type"] == user["property_type"]) &
    (filtered_df["bedrooms"] >= user["bedroom"]) &
    (filtered_df["bathrooms"] >= user["bathrooms"]) &
    (filtered_df["kitchen"] >= user["kitchen"]) &
    (filtered_df["wifi"] >= user["wifi"]) &
    (filtered_df["parking"] >= user["parking"]) &
    (filtered_df["ac"] >= user["ac"]) &
    (filtered_df["balcony"] >= user["balcony"]) &
    (filtered_df["self_check_in"] >= user["self_check_in"])
].copy()
```

```python
def calculate_score(property):

    #For Location Score
    if user["city"] == property["city"]:
        location_score = 0.20
    else:
        location_score = 0

    #For Budget Score
    price_difference = abs(user["budget"] - property["price_per_night"])
    budget_match = 1 - (price_difference / user["budget"])
    budget_score = max(0, budget_match) * 0.20

    #For Bedroom Score 
    if property["bedrooms"] >= user["bedroom"]:
        bedroom_score = 0.12
    else:
        bedroom_score = 0
        
    #Fo Bathroom Score
    if property["bathrooms"] >= user["bathrooms"]:
        bathroom_score = 0.08
    else:
        bathroom_score = 0


    #For Amenities Score
    amenities_score = 0

    if user["kitchen"] == 1 and property["kitchen"] == 1:
        amenities_score += 0.0375

    if user["wifi"] == 1 and property["wifi"] == 1:
        amenities_score += 0.025

    if user["parking"] == 1 and property["parking"] == 1:
        amenities_score += 0.075

    if user["ac"] == 1 and property["ac"] == 1:
        amenities_score += 0.0125


    #For Rating Score
    rating_score = (property["rating"] / 5) * 0.07
    max_reviews = df["review_count"].max()
    review_score = (property["review_count"] / max_reviews) * 0.03

    #For Verified Scoe
    if property["available"] == 1:
        availability_score = 0.05
    else:
        availability_score = 0

    if property["verified"] == 1:
        verified_score = 0.05
    else:
        verified_score = 0

    #For Preferences Score
    preferences = ["balcony", "terrace", "workspace","pool", "washing_machine"]

    matched_preferences = 0
    selected_preferences = 0

    for feature in preferences:
        if user[feature] == 1:
            selected_preferences += 1
            if property[feature] == 1:
                matched_preferences += 1

    preference_score = (
        matched_preferences / selected_preferences * 0.05
        if selected_preferences > 0 else 0
    )

    final_score = (
        location_score
        + budget_score
        + bedroom_score
        + bathroom_score
        + amenities_score
        + rating_score
        + review_score
        + availability_score
        + verified_score
        + preference_score
    )

    return final_score
```

```python
filtered_df["final_score"] = filtered_df.apply(calculate_score, axis=1)
```

```python
recommendations = filtered_df.sort_values("final_score",ascending=False)
```

```python
recommendations[["property_id",
                "property_name",
                "city",
                "property_type",
                "max_guests",
                "bedrooms",
                "beds",
                "bathrooms",
                "price_per_night",
                "price_3hr",
                "price_6hr",
                "price_9hr",
                "rating",
                "review_count",
                "verified",
                "self_check_in",
                "kitchen",
                "wifi",
                "parking",
                "ac",
                "balcony",
                "final_score",
                "verification_score",
                "verification_status"
]].head(10)
```

```python
top_property = recommendations.iloc[0]

reasons = []

if top_property["city"] == user["city"]:
    reasons.append("Preferred location")

if top_property["price_per_night"] <= user["budget"]:
    reasons.append("Within your budget")
else:
    reasons.append("Above your budget")

if top_property["bedrooms"] >= user["bedroom"]:
    reasons.append("Bedroom requirement matched")

if top_property["bathrooms"] >= user["bathrooms"]:
    reasons.append("Bathroom requirement matched")

if top_property["kitchen"] == 1 and user["kitchen"] == 1:
    reasons.append("Kitchen available")

if top_property["wifi"] == 1 and user["wifi"] == 1:
    reasons.append("Wi-Fi available")

if top_property["parking"] == 1 and user["parking"] == 1:
    reasons.append("Free parking available")

if top_property["ac"] == 1 and user["ac"] == 1:
    reasons.append("Air conditioning available")

if top_property["rating"] >= 4.5:
    reasons.append("Highly rated")

if top_property["review_count"] >= 50:
    reasons.append("Many guest reviews")

if top_property["verified"] == 1:
    reasons.append("Verified property")
```

```python
print("Property:", top_property["property_name"])
print("Match Score:", round(top_property["final_score"] * 100, 2))
print("Why this stay:")
for reason in reasons:
    print("✓", reason)
```
